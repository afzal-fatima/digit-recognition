```python

import os
import gdown
import torch
import requests
from torch.utils.data import Dataset
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt
from torchvision import datasets, models, transforms
from torch import nn, optim

# GitHub repository URL containing the images
github_repo_base_url = 'https://api.github.com/repos/Axzyl/ioai-day3/contents/'
folders = ['numbers', 'num_easy']

# Output directory to save downloaded images
output_base_dir = 'images'
os.makedirs(output_base_dir, exist_ok=True)

def download_images_from_github(github_repo_url, output_dir):
    response = requests.get(github_repo_url)
    files = response.json()

    for file_info in files:
        if file_info['name'].endswith(('png', 'jpg', 'jpeg')):
            file_url = file_info['download_url']
            response = requests.get(file_url)
            with open(os.path.join(output_dir, file_info['name']), 'wb') as f:
                f.write(response.content)

class CustomImageDataset(Dataset):
    def __init__(self, image_dir, transform=None):
        self.image_dir = image_dir
        self.transform = transform
        self.image_labels = []
        self.image_files = []

        for file_name in os.listdir(image_dir):
            if file_name.endswith(('png', 'jpg', 'jpeg')):
                label = int(file_name.split('-')[0])
                self.image_labels.append(label)
                self.image_files.append(file_name)

    def __len__(self):
        return len(self.image_files)

    def __getitem__(self, idx):
        img_path = os.path.join(self.image_dir, self.image_files[idx])
        image = Image.open(img_path).convert("RGB")
        label = self.image_labels[idx]

        if self.transform:
            image = self.transform(image)

        return image, label

datasets = {}
output_dirs = []
i = 0
for folder in folders:
    folder_url = f"{github_repo_base_url}{folder}"
    output_dir = os.path.join(output_base_dir, folder)
    os.makedirs(output_dir, exist_ok=True)

    # Download images from GitHub for the current folder
    download_images_from_github(folder_url, output_dir)

    # Create an instance of the dataset for the current folder
    output_dirs.append(output_dir)
    i += 1
```

```python
from torchvision import datasets, transforms
import torch

# Step 1: Calculate the mean and std for normalization
train_dataset = datasets.MNIST(root='./data', train=True, download=True, transform=transforms.ToTensor())

mean = train_dataset.data.float().mean() / 255
std = train_dataset.data.float().std() / 255

print(f"Mean: {mean}, Std: {std}")

mean = 0.1307
std = 0.3081



transform = transforms.Compose([
    transforms.RandomAffine(degrees=(0,20),translate=(0,0.2), scale=(0.7, 1)),
    transforms.Grayscale(num_output_channels=3),
    transforms.ToTensor(),
    transforms.Normalize((mean,), (std,))
])



# Don't modify these two lines - they download and initalize the dataset
dataset_1 = CustomImageDataset(image_dir=output_dir_1, transform=transform)
dataset_2 = CustomImageDataset(image_dir=output_dir_2, transform=transform)
```

```python
#creating data loaders
batch_size = 100

test_loader_1 = torch.utils.data.DataLoader(
    dataset_1,
    batch_size=batch_size,
    shuffle=True)

test_loader_2 = torch.utils.data.DataLoader(
    dataset_2,
    batch_size=batch_size,
    shuffle=True)

# Combine testing datasets
test_loader = torch.utils.data.DataLoader(torch.utils.data.ConcatDataset([dataset_1, dataset_2]), batch_size=batch_size, shuffle=True)

# Download and initialize the MNIST training dataset
train_transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((mean,), (std,))
])
train_dataset = datasets.MNIST(root='./data', train=True, download=True, transform=train_transform)
train_loader = torch.utils.data.DataLoader(train_dataset, batch_size=batch_size, shuffle=True)

# Defining the model
device = torch.device("cuda:0" if torch.cuda.is_available() else "cpu")
model = nn.Sequential(
    nn.Linear(28 * 28, 128),
    nn.Linear(128, 64),
    nn.Linear(64, 10)
).to(device)

loss_fn = nn.CrossEntropyLoss()
optimizer = optim.SGD(model.parameters(),lr=0.005)

#train the model
num_epochs = 10

for epoch in range(num_epochs):
    model.train()
    running_loss = 0.0

    for i, data in enumerate(train_loader):
        images, labels = data
        images = images.to(device)
        labels = labels.to(device)

        images = images.view(images.size(0), -1)  # Flatten the images
        pred = model(images)
        loss = loss_fn(pred, labels)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        running_loss += loss.item()

        if i % 100 == 99:
            print(f"epoch {epoch+1}, step {i+1}, loss = {loss.item()}")

    print(f"Epoch {epoch+1}/{num_epochs}, Loss: {running_loss/len(train_loader)}")

# Save the trained model to local storage
model_save_path = 'trained_model.pth'
torch.save(model.state_dict(), model_save_path)
print(f"Model saved to {model_save_path}")


  #Evaluate the Model
model.eval()
total_accuracy = 0
total_samples = 0

with torch.no_grad():
    for data in train_loader:
        images, labels = data
        images = images.to(device)
        labels = labels.to(device)

        images = images.view(images.size(0), -1)
        outputs = model(images)
        _, predicted = torch.max(outputs.data, 1)
        total_samples += labels.size(0)
        total_accuracy += (predicted == labels).sum().item()

accuracy = total_accuracy / total_samples
print(f"Accuracy: {accuracy:.4f}")

# Predict digit from an image file
image = Image.open("ADD FILE HERE")

model.eval()
preprocess = transforms.Compose([
    transforms.Resize((28, 28)),
    transforms.Grayscale(num_output_channels=1),
    transforms.ToTensor(),
    transforms.Normalize((mean,), (std,))
])

image_tensor = preprocess(image)
image_tensor = image_tensor.view(1, -1).to(device)

with torch.no_grad():
    output = model(image_tensor)
    predicted = torch.max(output, 1)[1].item()

print(f"Predicted class: {predicted}")

```
